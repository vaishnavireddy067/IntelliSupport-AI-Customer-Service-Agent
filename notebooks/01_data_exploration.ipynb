# 01: Customer Support on Twitter — AppleSupport Data Exploration

This notebook explores the AppleSupport conversation pairs extracted from the Kaggle *Customer Support on Twitter* dataset (`twcs.csv`).

### Goals:
1. Inspect dataset dimensions, conversation linkages, and schema.
2. Profile customer query length, vocabulary, and common issue themes.
3. Understand authentic customer noise: typos, slang, casing, and emojis.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is in sys.path
project_root = os.path.abspath('..')
if project_root not in sys.path:
    sys.path.insert(0, project_root)

data_path = os.path.join(project_root, 'data', 'processed', 'apple_conversations.csv')
if not os.path.exists(data_path):
    print(f'Data not found at {data_path}. Please run python scripts/prepare_data.py first.')
else:
    df = pd.read_csv(data_path)
    print(f'Successfully loaded {len(df):,} AppleSupport conversation pairs.')
    df.head(3)
```

## 1. Message Length Distribution

```python
if 'df' in locals():
    df['cust_char_len'] = df['customer_text'].fillna('').str.len()
    df['cust_word_len'] = df['customer_text'].fillna('').str.split().apply(len)
    
    print('Customer Query Length Summary (Characters):')
    print(df['cust_char_len'].describe())
    
    plt.figure(figsize=(10, 4))
    sns.histplot(df['cust_char_len'], bins=50, kde=True, color='royalblue')
    plt.title('Customer Message Character Length Distribution')
    plt.xlabel('Number of Characters')
    plt.ylabel('Count')
    plt.show()
```
